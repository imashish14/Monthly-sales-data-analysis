```python
#data load
import pandas as pd

df=pd.read_excel("C:/Users/Anubhav1Rana/OneDrive/Desktop/data analitycs (project)/sales_raw_500.xlsx")
```

```python
df
```

```python
#removing duplicate
df=df.drop_duplicates()
```

```python
df
```

```python
#snakecasing column names
df.columns=df.columns.str.lower()
df
```

```python
#fixing date column

df['order_date']=pd.to_datetime(df['order_date'], errors='coerce')
```

```python
df=df.dropna(subset=['order_date'])
```

```python
df
```

```python
#adding column 1.profit,2.year and 3.month

df['profit']=df['sales']-df['cost']
df['year']=df['order_date'].dt.year
df['month']=df['order_date'].dt.month
```

```python
df
```

```python
 df=df.drop(columns=['oder_date'])
```

```python
df.to_csv('sales_cleaned.csv',index=False)
```
